```python
# ============================================
# CELL 1 — Install libraries (Colab)
# ============================================
!pip -q install xgboost

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)
from sklearn.inspection import permutation_importance

pd.set_option("display.max_columns", 200)
```

```python
# ============================================
# CELL 2 — Load data (Upload method)
# ============================================
from google.colab import files
uploaded = files.upload()  # upload Train_set_(2).csv, Test_set_(2).csv, etc.

train_path = "Train_set_(2).csv"
test_path  = "Test_set_(2).csv"

train_df = pd.read_csv(train_path)
test_df  = pd.read_csv(test_path)

print("Train shape:", train_df.shape)
print("Test shape :", test_df.shape)
display(train_df.head())
```

```python
# ============================================
# CELL 4 — Basic EDA (lightweight but useful)
# ============================================
# Class balance
default = 'default' # Assuming the target column is named 'default'
y = train_df[default].astype(int)
default_rate = y.mean()
print(f"Default rate: {default_rate:.4f} ({default_rate*100:.2f}%)")

# Baseline accuracy if predict all zeros
baseline_acc = (y == 0).mean()
print(f"Baseline accuracy (predict all 0): {baseline_acc:.4f}")

# Missing values (top 20)
missing = train_df.isna().mean().sort_values(ascending=False)
print("\nTop missing columns:")
display(missing.head(20))

# Basic numeric summary
num_cols_preview = train_df.select_dtypes(include=[np.number]).columns.tolist()
print("\nNumeric columns count:", len(num_cols_preview))
```

```python
# ============================================
# CELL 5 — Split data (train/validation)
# ============================================
X = train_df.drop(columns=[default])
X_test_final = test_df.copy()

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Train split:", X_train.shape, "Valid split:", X_valid.shape)
```

```python
 #============================================
# CELL 6 — Build preprocessing pipeline
# ============================================
ID_COL = 'ID'
# Drop ID from features (keep it only for submission)
feature_cols = [c for c in X_train.columns if c != ID_COL]

X_train = X_train[feature_cols]
X_valid = X_valid[feature_cols]
X_test_final = X_test_final[feature_cols]

numeric_features = X_train.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = [c for c in feature_cols if c not in numeric_features]

print("Numeric features     :", len(numeric_features))
print("Categorical features :", len(categorical_features))

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler(with_mean=False))  # safe for sparse outputs
])
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=True))
])

preprocess = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
# ============================================
# CELL 7 — Define 3 models
# ============================================
# Helpful for imbalance
pos = (y_train == 1).sum()
neg = (y_train == 0).sum()
scale_pos_weight = neg / max(pos, 1)

models = {
    "LogReg": LogisticRegression(
        max_iter=3000,
        solver="saga",
        n_jobs=-1,
        class_weight="balanced"  # helps if defaults are rare
    ),
    "RandomForest": RandomForestClassifier(
        n_estimators=400,
        random_state=42,
        n_jobs=-1,
        class_weight="balanced_subsample"
    ),
    "XGBoost": XGBClassifier(
        n_estimators=800,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.9,
        colsample_bytree=0.9,
        reg_lambda=1.0,
        random_state=42,
        n_jobs=-1,
        eval_metric="logloss",
        scale_pos_weight=scale_pos_weight
    )
}
```

```python
# ============================================
# CELL 8 — Train & evaluate (Accuracy, Precision, Recall)
# ============================================
results = []

def evaluate_model(name, pipeline, Xv, yv):
    # Predict probabilities when available for threshold tuning
    if hasattr(pipeline.named_steps["model"], "predict_proba"):
        prob = pipeline.predict_proba(Xv)[:, 1]
    else:
        prob = None

    # Default threshold 0.5
    pred = pipeline.predict(Xv)

    acc = accuracy_score(yv, pred)
    prec = precision_score(yv, pred, zero_division=0)
    rec = recall_score(yv, pred, zero_division=0)
    f1 = f1_score(yv, pred, zero_division=0)

    return acc, prec, rec, f1, prob

trained_pipelines = {}
for name, model in models.items():
    pipe = Pipeline(steps=[
        ("preprocess", preprocess),
        ("model", model)
    ])

    pipe.fit(X_train, y_train)
    trained_pipelines[name] = pipe

    acc, prec, rec, f1, prob = evaluate_model(name, pipe, X_valid, y_valid)
    results.append([name, acc, prec, rec, f1])

results_df = pd.DataFrame(results, columns=["Model", "Accuracy", "Precision", "Recall", "F1"])
results_df = results_df.sort_values("Accuracy", ascending=False)
display(results_df)

best_model_name = results_df.iloc[0]["Model"]
best_pipe = trained_pipelines[best_model_name]
print("Best model by validation Accuracy:", best_model_name)

# Confusion matrix + report
pred_valid = best_pipe.predict(X_valid)
print("\nConfusion Matrix:\n", confusion_matrix(y_valid, pred_valid))
print("\nClassification Report:\n", classification_report(y_valid, pred_valid, zero_division=0))
```

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
import numpy as np
import pandas as pd

# 1) Base model (start from your winning setup)
xgb_base = XGBClassifier(
    objective="binary:logistic",
    eval_metric="logloss",
    tree_method="hist",          # faster on tabular + large data
    random_state=42,
    n_jobs=-1,
    scale_pos_weight=scale_pos_weight
)

pipe_xgb = Pipeline(steps=[
    ("preprocess", preprocess),
    ("model", xgb_base)
])

# 2) Parameter search space (targets generalization + accuracy)
param_dist = {
    "model__n_estimators": [600, 900, 1200, 1600],
    "model__learning_rate": [0.02, 0.03, 0.05, 0.07],
    "model__max_depth": [3, 4, 5, 6, 7],
    "model__min_child_weight": [1, 2, 5, 8, 12],
    "model__gamma": [0, 0.5, 1, 2, 4],
    "model__subsample": [0.7, 0.8, 0.9, 1.0],
    "model__colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    "model__reg_lambda": [0.5, 1.0, 2.0, 5.0, 10.0],
    "model__reg_alpha": [0.0, 0.1, 0.5, 1.0]
}

# 3) Cross-validation on TRAIN split only (avoid peeking at validation)
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

search = RandomizedSearchCV(
    estimator=pipe_xgb,
    param_distributions=param_dist,
    n_iter=30,                  # increase to 50+ if you have time
    scoring="accuracy",         # matches hackathon metric
    cv=cv,
    verbose=1,
    n_jobs=-1,
    random_state=42
)

search.fit(X_train, y_train)

print("\nBest CV Accuracy:", search.best_score_)
print("Best Params:")
best_params = search.best_params_
for k, v in best_params.items():
    print(f"  {k}: {v}")

# 4) Evaluate best model on your held-out VALIDATION split
best_pipe = search.best_estimator_
pred_valid = best_pipe.predict(X_valid)

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
acc = accuracy_score(y_valid, pred_valid)
prec = precision_score(y_valid, pred_valid, zero_division=0)
rec = recall_score(y_valid, pred_valid, zero_division=0)
f1 = f1_score(y_valid, pred_valid, zero_division=0)

print("\nValidation Metrics (tuned XGBoost):")
print(f"Accuracy : {acc:.6f}")
print(f"Precision: {prec:.6f}")
print(f"Recall   : {rec:.6f}")
print(f"F1       : {f1:.6f}")

print("\nConfusion Matrix:\n", confusion_matrix(y_valid, pred_valid))
print("\nClassification Report:\n", classification_report(y_valid, pred_valid, zero_division=0))
```

```python
import numpy as np
from sklearn.metrics import accuracy_score

probs = best_pipe.predict_proba(X_valid)[:, 1]
thresholds = np.linspace(0.1, 0.9, 81)

best_t, best_acc = 0.5, -1
for t in thresholds:
    pred_t = (probs >= t).astype(int)
    acc_t = accuracy_score(y_valid, pred_t)
    if acc_t > best_acc:
        best_acc = acc_t
        best_t = t

print(f"Best threshold for VALID accuracy: {best_t:.2f} with Accuracy={best_acc:.6f}")
```

```python
# ============================================
# CELL 10 — Feature importance (Permutation Importance)
# ============================================
# Note: This can be a bit slow on large data. We'll sample validation rows to speed it up.
sample_n = min(5000, X_valid.shape[0])
Xv_sample = X_valid.sample(sample_n, random_state=42)
yv_sample = y_valid.loc[Xv_sample.index]

print("Running permutation importance on sample size:", sample_n)
r = permutation_importance(
    best_pipe, Xv_sample, yv_sample,
    n_repeats=5,
    random_state=42,
    scoring="accuracy"
)

# Get expanded feature names after one-hot
ohe = best_pipe.named_steps["preprocess"].named_transformers_["cat"].named_steps["onehot"]
cat_feature_names = ohe.get_feature_names_out(categorical_features) if len(categorical_features) else np.array([])
# The original all_feature_names was constructed using the expanded cat_feature_names,
# which resulted in a length mismatch with r.importances_mean.
# Permutation importance is calculated on the original features passed to the pipeline.
# So, we should use the columns of the input sample as the index.
# all_feature_names = np.concatenate([np.array(numeric_features), cat_feature_names])

importances = pd.Series(r.importances_mean, index=Xv_sample.columns).sort_values(ascending=False)
top_k = 20
display(importances.head(top_k))
plt.figure(figsize=(10, 6))
importances.head(top_k).sort_values().plot(kind="barh")
plt.title(f"Top {top_k} Feature Importances (Permutation, Accuracy impact)")
plt.xlabel("Mean importance")
plt.tight_layout()
plt.show()
```

```python
# ============================================
# CELL 11 — Train final model on FULL train and predict test
# ============================================
# Refit best model on entire training data
X_full = train_df.drop(columns=[default])
y_full = train_df[default].astype(int)

X_full = X_full[[c for c in X_full.columns if c != ID_COL]]
X_test_for_pred = test_df[[c for c in test_df.columns if c != ID_COL]]

final_pipe = Pipeline(steps=[
    ("preprocess", preprocess),
    ("model", models[best_model_name])  # same model type/params
])

final_pipe.fit(X_full, y_full)

# Predict using tuned threshold if possible
if hasattr(final_pipe.named_steps["model"], "predict_proba"):
    test_prob = final_pipe.predict_proba(X_test_for_pred)[:, 1]
    test_pred = (test_prob >= best_t).astype(int)
else:
    test_pred = final_pipe.predict(X_test_for_pred)
```

```python
# ============================================
# CELL 12 — Create submission file
# ============================================
submission = pd.DataFrame({
    "ID": test_df[ID_COL],
    "default": test_pred.astype(int)
})

print("Submission shape:", submission.shape)
display(submission.head())

out_path = "submission.csv"
submission.to_csv(out_path, index=False)
print("Saved:", out_path)

# Download to your computer (optional)
files.download(out_path)
```
